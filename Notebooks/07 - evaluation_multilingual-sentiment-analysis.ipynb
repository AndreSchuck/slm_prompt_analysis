# Multilingual Sentiment Classification Model Evaluation
This Jupyter notebook implements an evaluation pipeline for the [Multilingual-sentiment-analysis](https://huggingface.co/tabularisai/multilingual-sentiment-analysis) language model on Portuguese language sentiment analysis tasks. The notebook is designed to:



1.   Systematically evaluate the model's performance across multiple Brazilian Portuguese datasets (IMDB-PT, OpCovidBR, RELI, and TweetSentBR);
2.   Test different prompting strategies using various templates for few-shot and zero-shot learning;
3. Conduct multiple evaluation rounds (3 rounds);
4. Track inference time and model predictions;
5. Handle large-scale evaluations with checkpoint saving and recovery mechanisms.


The notebook is designed to run locally or in any Jupyter-compatible environment with GPU support and requires environment variables defined in a `.env` file (see `.env.example` at the project root). This evaluation is part of a broader assessment of language models' capabilities in Portuguese language tasks.

# 03 - Functions

Definition of the functions that will be used during predictions.

````python
def get_demonstrations_df(dataset: pd.DataFrame, random_state: int = 42) -> pd.DataFrame:
    """
    Randomly shuffle a dataset for demonstration purposes.
    Args:
        dataset (pd.DataFrame): The input DataFrame to sample from.
        random_state (int, optional): Random seed for reproducibility. Defaults to 42.
    Returns:
        pd.DataFrame: The shuffled DataFrame values.
    """
    demo_df = dataset.sample(frac=1, random_state=random_state).values
    return demo_df

def generate_staging_area_csv_file_name(model_name: str, dataset_name: str, first_index: int, last_index: int) -> str:
    """
    Generate a standardized CSV file name for the staging area.
    Args:
        model_name (str): Name of the model.
        dataset_name (str): Name of the dataset.
        first_index (int): Starting index of the data range.
        last_index (int): Ending index of the data range.
    Returns:
        str: Formatted file name string following the pattern 'model_dataset_prompt_round_start-end.csv'.
    """
    file_name = f'{model_name}_{dataset_name}__{first_index}-{last_index}.csv'
    return file_name

def get_staging_area_files_by_model_or_dataset_name(model_name: str = None, dataset_name: str = None) -> list[Path]:
    """
    Retrieve sorted list of staging area files matching specified criteria.
    Args:
        model_name (str, optional): Model name to filter by. Defaults to None.
        dataset_name (str, optional): Dataset name to filter by. Defaults to None.
    Returns:
        list[Path]: Sorted list of Path objects for matching CSV files.
    """
    files = [csv_file for csv_file in STAGING_AREA_ROOT_PATH.glob(f'{model_name}_{dataset_name}*.csv')]
    return sorted(files, key=lambda x: int((str(x.absolute()).split('-')[-1]).split('.')[0]))

def get_last_staging_area_file_content_index(model_name: str = None, dataset_name: str = None) -> int | None:
    """
    Get the last index from staging area files for specified criteria.
    Args:
        model_name (str, optional): Model name to filter by. Defaults to None.
        dataset_name (str, optional): Dataset name to filter by. Defaults to None.
    Returns:
        int: Last index found in matching files, or None if no files found.
    """
    files = get_staging_area_files_by_model_or_dataset_name(model_name=model_name, dataset_name=dataset_name)
    if len(files) == 0:
        return None
    last_file = str(files[-1].absolute())
    last_file_index_name = int((last_file.split('_')[-1]).split('.')[0].split('-')[-1])
    return last_file_index_name

def save_csv_to_staging_area(model_name: str, dataset_name: str, results_list: list) -> Path:
    """
    Save experiment results to a CSV file in the staging area.
    Args:
        model_name (str): Name of the model.
        dataset_name (str): Name of the dataset.
        results_list (list): List of tuples containing (index, prediction, time) results.
    Returns:
        Path: Path to the saved CSV file.
    """
    sorted_result_list = sorted(results_list, key=lambda x: x[0])
    first_index = sorted_result_list[0][0]
    last_index = sorted_result_list[-1][0]
    file_name = generate_staging_area_csv_file_name(
        model_name=model_name,
        dataset_name=dataset_name,
        first_index=first_index,
        last_index=last_index
        )

    file_path = STAGING_AREA_ROOT_PATH / file_name
    df = pd.DataFrame(results_list, columns=['index','predictions', 'time'])
    df.to_csv(file_path)
    print(f'File containing index {first_index} to {last_index} for  model {model_name}, and dataset {dataset_name} created at{file_path} with success.')
    return file_path.parent / (file_path.name +'.csv')

def convert_csv_in_staging_area_to_dataframe(list_of_files: list[Path]) -> pd.DataFrame:
    """
    Combine multiple staging area CSV files into a single sorted DataFrame.
    Args:
        list_of_files (list[Path]): List of paths to CSV files in the staging area.
    Returns:
        pd.DataFrame: Combined and sorted DataFrame with predictions and timing information.
    """
    df = pd.DataFrame(columns=["Unmatch: 0", "index", "predictions", "time"])
    for file in list_of_files:
        temp_df = pd.read_csv(file)
        df = pd.concat([df, temp_df])
    df = df.sort_values(by='index')
    df = df.drop_duplicates(subset=['index'])
    df = df.reset_index(drop=True).set_index('index')
    df = df[['predictions', 'time']]
    return df

def process_sentiment_predictions(prediction_scores: list[dict], task_labels: list[tuple]) -> int:
    """
    Processes sentiment prediction scores based on task type (binary or multi-class).

    Args:
        prediction_scores (list[dict]): A list of dictionaries, where each dictionary
                                        contains 'label' (e.g., 'Negative', 'Positive', 'Neutral',
                                        'Very Negative', 'Very Positive') and its associated 'score'.
        task_labels (list[tuple]): A list of tuples defining the possible labels for the task.
                                   E.g., [(-1, 'Negativo'), (1, 'Positivo')] for binary,
                                   or [(-1, 'Negativo'), (0, 'Neutro'), (1, 'Positivo')] for multi-class.

    Returns:
        int: The predicted sentiment label (-1, 0, or 1).
    """
    negative_score = 0.0
    positive_score = 0.0
    neutral_score = 0.0

    # Aggregate scores based on sentiment categories
    for pred in prediction_scores:
        label = pred['label']
        score = pred['score']
        if label in ['Negative', 'Very Negative']:
            negative_score += score
        elif label in ['Positive', 'Very Positive']:
            positive_score += score
        elif label == 'Neutral':
            neutral_score += score

    # Determine task type
    if len(task_labels) == 2:  # Binary classification
        # Disregard neutral_score for binary classification
        if negative_score >= positive_score: # Tie-breaking: prefer negative if scores are equal
            return -1
        else:
            return 1
    elif len(task_labels) == 3:  # Multi-class classification
        scores = {
            -1: negative_score,
            0: neutral_score,
            1: positive_score
        }
        # Find the label with the highest score
        return max(scores, key=scores.get)
    else:
        raise ValueError("Unsupported task_labels length. Expected 2 (binary) or 3 (multi-class).")
````

## 04 - Multilingual-sentiment-analysis Evaluation

## 04.01 - Load model

````python
pipe = pipeline(
    "text-classification",
    model="tabularisai/multilingual-sentiment-analysis",
    truncation=True,
    device="cuda")
````

# 04.02 - Perform inference

Perform inference for all datasets and prompt templates for 3 rounds.

````python
for df_name, df in TEST_DATASETS.items():
    task_labels = TASK_LABELS.get(DATASETS_TASK_TYPES.get(df_name))
    # Check if this combination has already been evaluated
    if f"{MODEL_NAME}_{df_name}" not in RESULT_DATASETS:
        print(f'Starting evaluation on dataset: {df_name}')
        results_list = []
        time_list = []
        # Find the last processed index from previous runs (if any)
        first_index = get_last_staging_area_file_content_index(dataset_name=df_name,                              model_name=MODEL_NAME)
        # Start from beginning if no previous files found
        if first_index == None:
            first_index = 0
            print(f'No file were found for the current dataset and model, starting from index 0')
        # Otherwise, continue from last processed index
        else:
            first_index += 1
            print(f'One or more files were found. Starting from lastest index found ({first_index}) for dataset {df_name}')

        # Process each text item in the dataset starting from first_index
        for index, item in df['df']['text'][first_index:].items():
            t0 = time()
            outputs = pipe(item, top_k=5)
            predicted_label = process_sentiment_predictions(outputs, task_labels)
            results_list.append((index, predicted_label, time()-t0))

            # Save checkpoint every 100 predictions
            if len(results_list) % 100 == 0:
                save_csv_to_staging_area(model_name=MODEL_NAME,
                                         dataset_name=df_name,results_list=results_list)

        # Save final checkpoint after all predictions
        save_csv_to_staging_area(
            model_name=MODEL_NAME,
            dataset_name=df_name,
            results_list=results_list)


        # Handle cases where results might be split across multiple files
        if len(results_list) != (df['df'].shape[0]):

                # Combine all staging area files for this configuration
            df_temp = convert_csv_in_staging_area_to_dataframe(
                get_staging_area_files_by_model_or_dataset_name(
                    dataset_name=df_name,
                    model_name=MODEL_NAME))

            # If combined results match dataset size, update the dataframe
            if df_temp.shape[0] == df['df'].shape[0]:
                df['df']['predictions'] = df_temp['predictions']
                df['df']['time'] = df_temp["time"]

        # Handle case where all results are in single results_list
        elif len(results_list) == (df['df'].shape[0]):
            df['df']['predictions'] = np.array(results_list)[:,1]
            df['df']['time'] = np.array(results_list)[:,2]

            # Raise error if results count doesn't match dataset size
        else:
            raise ValueError('The number of results is different from the number of rows in the dataset.')

            # Save final results to CSV

        (RESULTS_DATA_ROOT_PATH / MODEL_NAME).mkdir(parents=True, exist_ok=True)

        df['df'].to_csv(f'{RESULTS_DATA_ROOT_PATH}/{MODEL_NAME}/{MODEL_NAME}_{df_name}_result.csv', index=False)

            # Update list of completed evaluations
        RESULT_DATASETS = [dataset.stem.split("_result")[0] for dataset in RESULTS_DATA_ROOT_PATH.glob(f'{MODEL_NAME}/*.csv')]
        print(f'The evaluation of dataset: {df_name} has ended')
````
